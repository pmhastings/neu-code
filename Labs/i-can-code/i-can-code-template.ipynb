## Array Creation

Unlike the previous results which were single numbers (also called *scalars*), with `numpy`, we can easily create matrices and vectors, i.e., structures collections of numbers. An array can have any number of dimensions. An array with either only one row or one column is referred to as a vector. You can find more information at this link: [numpy arrays and vectors](https://www2.lawrence.edu/fast/GREGGJ/Python/numpy/numpyLA.html). (*Matlab-Colab-diff: Creating arrays in numpy can be a little more cumbersome than in matlab, but it's got lots of options!*)

Examples of creating basic vectors or arrays:

1x4 row vector. Use spaces or commas between elements.

* `a = np.array([1, 2, 3, 4])`

4x1 column vector. **Use semicolons between rows.**

* `b = np.array([[1], [8], [2], [6]])`

4x3 array:

* `c = np.array([[1, 3, 5], [2, 4, 6], [7, 8, 10], [9, 12, 0]])`

<h3>Array Indexing</h3>

To access particular elements of an array, we use indexing. Each element within an array is positioned within a particular row and column; we can use these positions to manipulate specific elements. For more information, see this link: **MathWorks Array Indexing**

Examples:

Access the element in the 2nd row and 3rd column from array `c`, and assign it to the variable, x. *Matlab-Colab-diff: indices in numpy start with 0, not 1 like in matlab. And indexing is done with square brackets instead of parentheses.*

* `c = np.array([[1, 3, 5], [2, 4, 6], [7, 8, 10], [9, 12, 0]])`

* `x = c[1, 2]`

Change the value of the element in the 1st row and 1st column of array `c` to 18.

* `c[0, 0] = 18`

Access the entire 3rd row, i.e the 3rd row and all columns, of array c and assign it to the variable, `x`.

* `x = c[2,:]`

Access the entire 2nd column, i.e the 2nd column and all rows, of array `c` and assign it to the variable, `y`.

* `y = c[:,1]`

````python
# Example code (there is no coding for you to complete here).

# Matlab colab diff:
# Matlab instructions work in a particular context.
# matplotlib (in colab) is "object-oriented".
# The subplot command below makes objects for the figure (which we won't use now),
# and axes. Then we call methods of the axes objects with appropriate parameters.

xvals = np.arange(2)     # This will be an array [0, 1]
MEAN_data = [0.2, 0.6]   # These are the mean values you want to plot.
SEM_data = [0.03, 0.07]  # These are the standard error values associated with the mean

# This makes a new plot (empty for now), and returns its figure and axes.
# We don't need the figure part of this.
fig, ax = plt.subplots()
bar_width = .6           # width of the bars

# This line generates a bar plot using the values in the MEAN_data vector
ax.bar(xvals, MEAN_data, bar_width)

# This command adds error bars to the bar plot. The errorbars will be at the
# X values 1 and 2, at the Y values of MEAN_data, using the length from
# SEM_data.
# The fmt parameter indicates a blac(k) line.
# The underscore (_) before the k says NOT to draw a line connecting the
# error bar values.
# capsize gives the width of the horizontal part at the ends of the lines.
ax.errorbar(xvals, MEAN_data, yerr=SEM_data, fmt='_k', capsize=4)

# Tell the X axis where the ticks are and what they're called.
# matlab colab diff: in matlab, can use curly brackets {} for a list.
# in colab (python), {} gives a set, which is unordered
ax.set_xticks(xvals, ['Group 1', 'Group 2'])

ax.set_xlabel('X axis label')             # This labels the x-axis as 'x'.
ax.set_ylabel('Y axis label')             # This labels the y-axis as 'y'.
ax.set_ylim([0, 0.8])          # This sets the limits of the y-axis from 0 to 0.8
ax.set_title('Example code')   # This adds a title to the plot

plt.show()                     # Display the completed plot
````
